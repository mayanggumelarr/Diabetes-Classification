```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA, SparsePCA
from sklearn.model_selection import StratifiedKFold, KFold, cross_val_score
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("train.csv")
df.info()
```

**FEATURE EXPLANATION (Unknown Features Background) :** <br>

1. `ethnicity` : etnis. beberapa etnis memiliki risiko lebih tinggi
2. `education_level_&_income_level` : korelasi dengan akses gaya hidup sehat
3. `waist_to_hip_ration` : lebih akurat dari BMI karena mengetahui lemak perut yang sangat berkaitan dengan resistensi insulin
4. `systolic & diastolic` : tensi. Hipertensi sering terjadi dengan diabetes (komorbiditas)
5. `heart_rate` : mencerminkan kesehatan/keaktivan jantung
6. `cholesterol_total, hdl, ldl, triglycerides` : kondisi lemak darah. HDL (lemak baik) dan LDL (lemak jahat). penderita sering ber-hdl rendah trigliserida tinggi.
7. `screen_time_hours_per_day` : screen time bisa menjadi acuan penderita manusia aktif bergerak, kurang, atau bahkan tidak.

<br>

Ternyata, terkait kolestrol. Orang dengan LDL - Trigliserida tinggi namun HDL rendah, lebih berisiko terkena diabet. (Dislipidemia diabetik). Ini ada kaitannya dengan konsumsi makanan, aktivitas, dan normal tidaknya sekresi insulin (enzim pemicu sel serap glukosa)

```python
## Correlation
num_feat = df.select_dtypes(include='number')

plt.figure(figsize=(15,10))
corre = num_feat.corr()
sns.heatmap(corre, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

**PENJELASAN :**
- Ldl X Cholestrol Total = 81% (High with Risk)
- BMI X Waist Hip Ratio = 76% (Moderate High)
<br>

Dari hasil di atas, beberapa hal yang dipertimbangkan: <br>
1. Gunakan mutual information untuk mengetahui presentase mutual antara Ldl dan Cholestrol dg Target. Yang lebih kecil drop
2. Gunakan juga untuk BMI dan WHR. Lalu drop salah satu dg MI terkecil jika model tidak seimbang

```python
## Outlier Checking
plt.figure(figsize=(12, 10))
for i, col in enumerate(num_feat.columns, 1):
    plt.subplot(5, 5, i)
    sns.boxplot(x=num_feat[col])
    plt.title(f"{col}")
plt.tight_layout()
plt.show()
```

**PENJELASAN :**

Karena outlier sangat2 extrem, maka hindari pakai StandardScaler yang sensitif outlier. Then use RobustScaling (Mean & IQR focused)
